# Global 90/60: Return Stacking, Realistic Costs and the Limits of Reconstruction

## 1. Research question and intended output

Does a practically implementable global **90% equity / 60% government-bond** strategy offer an attractive net return–risk trade-off against 100% developed-market equities and a conventional 60/40 portfolio?

The deliverables are a reproducible GitHub analysis and a technical, quantitative blog post: clear enough for an informed investor, with the accounting, sources and limitations exposed. This is not a claim that leverage creates free diversification, or that a synthetic history is historical NTSG performance.

The investigation separates three evidence tiers:

1. **Primary long-history model:** an NTSG-inspired, financed cash-index 90/60 proxy using identical sleeves across strategies, isolating leverage and financing.
2. **Friction scenarios:** explicit financing/basis, roll, execution and margin-stress assumptions applied to that proxy; these model implementation risk, rather than reconstructing unobserved historical contract P&L.
3. **Live evidence:** actual NTSG returns from November 2024, with official NAV preferred and exchange prices used only as an explicitly labelled diagnostic.

The primary 1985–2026 proxy uses MSCI World net total return, FTSE WGBI Developed Markets USD-unhedged total return and USD bill returns. It deliberately does not reproduce NTSG's equity screens, exact eight-contract government-futures basket, currency-weighted multicurrency collateral, or issuer-specific reset rules. It must therefore be named **NTSG-inspired 90/60 proxy**, never historical NTSG performance. The sample will follow verified coverage; missing history will not be manufactured or silently spliced. Pareto dominance is a possible empirical finding under predefined metrics, not the research premise.

## 2. Why 90/60 is not automatically superior to 60/40

For a controlled, period-start rebalanced experiment:

$$
R_{60/40,t}=0.60R_{E,t}+0.40R_{B,t}
$$

$$
R_{90/60,t}=0.90R_{E,t}+0.60R_{B,t}-0.50R_{f,t}-D_t
$$

Hence:

$$
R_{90/60,t}-R_{f,t}=1.5\left(R_{60/40,t}-R_{f,t}\right)-D_t
$$

Before costs, excess returns scale by 1.5; their Sharpe ratio is unchanged. With constant financing rates, ordinary return volatility also scales by 1.5. A constant per-period cost lowers the Sharpe ratio. These are periodic arithmetic-return identities, not identities for CAGR or drawdown. Currency translation, drifting weights, different instruments and implementation frictions must be evaluated separately.

The theoretical foundation is risky-portfolio scaling with a financing asset, not a new efficient frontier; see [Tobin (1958), Liquidity Preference as Behavior Towards Risk](https://doi.org/10.2307/2296205). The identities above expose the assumptions directly rather than relying on a claim of free capital efficiency.

| Strategy | Controlled exposure |
| --- | --- |
| 100% equities | 100% of the selected developed-market equity sleeve |
| 60/40 | 60% equities and 40% of the selected government-bond sleeve |
| 90/60 | 90% equities and 60% bonds, with financing accounted for once |

Use the same sleeve definitions, dates and rebalance convention within each controlled comparison. Also show an idealized 1.5× financed 60/40 accounting control and risk-matched comparisons. Any historical risk scaling must use lagged estimates; full-sample volatility matching is descriptive, not an investable strategy.

## 3. What NTSG actually implements

The reference is the [WisdomTree Global Efficient Core Index methodology, October 2025](https://www.wisdomtree.com/eu/-/media/eu-media-files/other-documents/research/index/wisdomtree-global-efficient-core-index-methodology.pdf), together with the [fund factsheet](https://dataspanapi.wisdomtree.com/pdr/documents/FACTSHEET/UCITS/EU/EN-GB/IE00077IIPQ8/).

- **Funded assets:** approximately 90% physical developed-market equities and 10% cash collateral. The bond overlay adds 60% futures notional: 150% risky-asset exposure, not 150% cash invested.
- **Equities:** an eligible developed-market universe, selection from the top 1,500 companies, ESG exclusions, free-float weights, liquidity adjustments and a 10% single-stock cap. Equity constituents are rebalanced annually in December. MSCI World is a broad proxy, not this exact basket.
- **Bonds:** eight government-bond futures referencing the US, Germany, Japan and UK, spanning 2–30-year maturity segments. Currency-group weights follow the corresponding equity currency weights after rescaling; selected contracts are equally weighted within each currency group. The exact historical contract mapping and weights must be sourced before claiming replication.
- **Collateral:** USD, EUR, GBP and JPY cash, with currency weights matching the bond sleeve. USD fund denomination does not establish blanket USD hedging of the bond overlay. Foreign futures also do not create FX exposure equal to holding their full notional in foreign cash bonds.
- **Scheduled resets:** last business day of February, May, August and November. Futures roll from the first-near to the second-near contract on the specified one-day quarterly roll dates.
- **Exceptional resets:** the methodology specifies a greater-than-5% exposure-deviation trigger. Confirm exact threshold semantics and execution timing before claiming faithful replication; a five-percentage-point interpretation must be labelled an assumption until verified.
- **Fund expenses:** current TER is 0.25% annually. Historical fee schedules, taxes within the fund, cash remuneration and tracking differences need separate treatment.

Applying today's published rules before the fund existed is a retrospective synthetic experiment. It is not evidence that this exact product or rule set was investable then.

### Two accounting models, never mixed

**Model A — financed cash-index approximation.** Use the equation in Section 2 only when $R_B$ is a cash-bond total return and $R_f$ is a properly constructed interval financing return. It assumes matched borrowing/collateral rates and an approximation between cash bonds and futures. A USD-hedged bond index is a possible controlled proxy, not an exact NTSG bond sleeve.

**Model B — physical equities, collateral and futures.** This is the target accounting framework for a future contract-level reconstruction; it cannot be treated as historical NTSG without the required contracts, weights and collateral data. The self-financing return is:

$$
R_t=w_{E,t-1}R_{E,t}+\sum_c w_{C_c,t-1}R^{\mathrm{USD}}_{C_c,t}+\frac{\sum_j\mathrm{PnL}^{\mathrm{USD}}_{j,t}}{\mathrm{NAV}_{t-1}}-C_t
$$

Here collateral returns include local interest and FX; futures P&L uses actual contract quantities, multipliers, settlement-price changes and currency conversion. Between resets, holdings and exposures drift. Transfers between cash and margin do not themselves create profit or loss. Fractional contract exposure is an institutional-scale approximation; smaller-account contract rounding belongs in sensitivity analysis.

Match and report duration/DV01 and yield-curve buckets, not only bond notional. Treasury futures track their cheapest-to-deliver securities and delivery economics; a broad cash-bond index can have materially different rate sensitivity. See CME's [cheapest-to-deliver explanation](https://www.cmegroup.com/education/courses/introduction-to-treasuries/get-to-know-treasuries-ctd) and [conversion-factor specification](https://www.cmegroup.com/articles/2024/calculating-us-treasury-futures-conversion-factors.html).

The bridge between models is an assumption to validate: with one cash rate and futures excess return approximated by $R_B-R_f$, the funded structure gives $0.90R_E+0.10R_f+0.60(R_B-R_f)=0.90R_E+0.60R_B-0.50R_f$. Multicurrency collateral, basis, delivery options and duration differences break exact equivalence.

Do **not** subtract another 50% financing charge from actual futures P&L. Do **not** add collateral income again to a futures series already defined as fully collateralized total return. Continuous-futures percentage changes are not accepted as contract-level P&L without verified stitching and roll rules.

### Costs

Separate fund TER, financing/basis, execution spread, commissions and roll-related economic return. The earlier flat 0.05% annual “roll cost” is not an established estimate and is removed as a factual baseline. Use documented execution assumptions and an explicitly labelled cost grid instead. Roll return and financing may already be embedded in a source series.

### Proxy implementation-friction model

The long-history proxy will include a transparent scenario cost, not invented historical futures P&L:

$$
R_{90/60,t}^{\mathrm{net}}=0.90R_{E,t}+0.60R_{B,t}-0.50R_{f,t}-C_{\mathrm{basis},t}-C_{\mathrm{roll},t}-C_{\mathrm{trade},t}-C_{\mathrm{fund},t}
$$

- $C_{\mathrm{basis},t}$ is a stated annualized financing/basis spread applied to the 60% futures notional; report a zero-cost accounting control and a predeclared scenario grid.
- $C_{\mathrm{roll},t}$ is a stated per-roll notional cost, charged only on scheduled rolls; it must not duplicate a roll return already contained in $R_B$.
- $C_{\mathrm{trade},t}$ is a stated bid–ask plus commission assumption times the observable proxy turnover at scheduled resets and exceptional-rebalance scenarios.
- $C_{\mathrm{fund},t}$ is either the 0.25% NTSG TER for an NTSG-cost proxy or zero for a pure index exposure; state the choice once and never double-count ETF expenses.

Margin is a **stress test**, not an inferred historical result: specify initial-margin fraction, variation-margin cash calls, a collateral floor and a forced-deleveraging rule. Monthly data can test month-end collateral sufficiency only; it cannot establish intra-month margin survival. Daily settlement data is required before reporting a margin-call frequency or a liquidation path.

ETF adjusted closes already include fund expenses and distributions: do not deduct their TER again. A synthetic index-based fund model may deduct its target fee once. Replacing an ETF's embedded fee requires a documented approximation, not an automatic add-back of today's TER. Apply comparable implementation-cost treatment to benchmarks. Exclude investor taxes; disclose fund-level withholding assumptions. A baseline without forced liquidation must not be described as a validated margin-survival simulation.

## 5. Currency, timing and data policy

For $S_t$ quoted in USD per EUR:

$$
1+R_{\mathrm{EUR},t}=(1+R_{\mathrm{USD},t})\frac{S_{t-1}}{S_t}
$$

Translate each period, then compound; translating every NAV observation is equivalent. Normalize each currency's starting NAV consistently. USD results use a USD cash reference; EUR Sharpe/Sortino calculations use a suitable EUR cash reference and a stated minimum acceptable return. Neither USD denomination nor EUR trading currency means EUR hedging.

Use daily accounting when the data supports rolls, threshold resets and margin diagnostics; report monthly results and closed-month validation. London ETF closes, US closes and Yahoo FX timestamps are not synchronous. Compare monthly/weekly diagnostics and use official NAV where possible rather than interpreting daily tracking noise as replication failure. Monthly-only data supports month-end drawdown, not exact daily maximum drawdown.

No backward filling, interpolated market prices, zero-filled missing returns or unlabelled pre-inception ETF extensions. Preserve source calendars. Any bounded holiday carry-forward must be explicit and must not conceal stale prices. Use common verified start dates, exclude incomplete terminal months, and never bridge missing-data gaps as if they were single-period returns. Rates used for accrual must be available at the interval start; quoted yields need a documented day-count and holding-period conversion.

1. **Freeze the experiment:** sample endpoints, primary comparisons, rebalance conventions, cost scenarios, financing sources, regime definitions and return/risk metrics before inspecting outcomes. Use monthly resets for the generic exposure experiment; issuer-specific resets for the NTSG model, with matched benchmark schedules in each panel.
2. **Audit the inputs:** verify total-return definitions, historical index membership/methodology, currency, units, distribution adjustments, source licensing and backfilled-versus-live periods. Acquire the missing index/contract data before presenting long-history results. Raw data redistribution on GitHub requires provider permission; otherwise publish code, identifiers and permitted derived outputs.
3. **Implement the controlled model:** verify the leveraged-60/40 identity, then add documented financing and costs. ETF-only tests remain labelled sensitivity experiments, with net-of-fee inputs and proxy mismatches visible.
4. **Implement the contract model only after its data gate passes:** historical futures settlements, exact rolls, multipliers, lagged position sizing, multicurrency collateral, reset triggers, transaction costs and margin diagnostics. Use historical collateral rates—not a single USD rate for all currencies.
5. **Reconcile accounting:** each NAV change must equal equity return, collateral interest/FX, futures P&L and costs. Test flat markets, zero rates, pure FX moves, fee-only paths, roll dates and zero-turnover periods. Resets and margin transfers must not generate wealth.
6. **Validate against live evidence:** compare with official NTSG index/fund NAV when obtained; use the downloaded USD listing as a secondary check. Report monthly correlation, annualized active return, tracking error and rolling differences over an explicit common sample. Do not tune proxy weights to the same short validation window; disclose its limited statistical power.
7. **Test robustness:** alternate start dates, duration/country exposures, financing spreads, costs, resets, FX assumptions and historical subperiods. Include the global financial crisis and 2022 only when the relevant sample covers them. A modern sample does not test 1970s stagflation.
8. **Quantify uncertainty:** paired block-bootstrap intervals preserving cross-asset dependence, with block-length sensitivity. Report uncertainty in return/risk differences; repeated regimes, metrics and parameter searches are exploratory. A holdout selected after the strategy became known is not genuine prospective evidence.

## 8. Mathematical anatomy of a 90/60 portfolio

This section distinguishes a **financed cash-index proxy** from a contract-level futures account. The former is the long-history model; the latter explains the mechanics that the proxy approximates. Neither turns the proxy into historical NTSG performance.

### 8.1 Capital, notional exposure and leverage

Start with NAV $V_{t-1}=100$. At a reset date, a 90/60 portfolio targets:

| Component | Capital allocation | Risky notional |
| --- | ---: | ---: |
| Developed equities | $90$ | $90$ |
| Cash collateral | $10$ | $0$ |
| Government-bond futures | margin only | $60$ |

The portfolio therefore has $150$ of risky notional exposure for $100$ of NAV:

$$
G = 0.90 + 0.60 = 1.50
$$

The extra $50$ is not borrowed cash invested in bonds. It is a $60$ futures overlay supported by the $10$ collateral reserve and the funded equity sleeve. The economic leverage relative to NAV is $G-1=0.50$. Margin is a performance bond, not the cost of the futures position; futures gains and losses are settled through variation margin.

### 8.2 Exact self-financing accounting

Let $S_t$ be the equity price, $F_{j,t}$ the settlement price of futures contract $j$, $M_j$ its contract multiplier, $X_{j,t}$ its USD conversion rate, $q_{j,t-1}$ the number of contracts held through period $t$, and $C_t$ the cash balance. With equity units $n_{E,t-1}$, the one-period NAV change is:

$$
V_t = V_{t-1}
+ n_{E,t-1}(S_t-S_{t-1})
+ C_{t-1}R_{C,t}
+ \sum_j q_{j,t-1}M_j(F_{j,t}-F_{j,t-1})X_{j,t}
- \mathrm{Cost}_t
$$

A cash transfer to initial margin does not change $V_t$: it merely moves wealth between cash and margin accounts. Variation margin changes cash but is already the futures profit and loss in the equation. If margin causes forced trading, the forced trade and its costs change later exposures and therefore future returns.

At a reset, target units for an institutional fractional-contract model are:

$$
n_{E,t}=\frac{0.90V_t}{S_t},
\qquad
q_{j,t}=\frac{0.60V_t\,\omega_{j,t}}{M_jF_{j,t}X_{j,t}},
\qquad
\sum_j\omega_{j,t}=1
$$

Between resets, do not reset weights mechanically: equity capital and futures notional drift with markets. Contract rounding is a separate small-account implementation sensitivity.

### 8.3 Cash-index proxy and the $-0.50R_f$ term

Let $R_E$ be the equity total return, $R_B$ a cash-government-bond total return, and $R_f$ the cash or financing return over the same period. If the futures overlay's **excess** return is approximated by:

$$
R_{B}^{\mathrm{excess}} \approx R_B-R_f
$$

then the funded structure has a 90% equity sleeve, 10% cash collateral and 60% futures excess exposure:

$$
R_{90/60}^{\mathrm{gross}}
= 0.90R_E + 0.10R_f + 0.60(R_B-R_f)
= 0.90R_E + 0.60R_B - 0.50R_f
$$

The $-0.50R_f$ term is therefore an accounting consequence of expressing a fully funded bond total return as a futures excess return while only 10% of NAV earns cash. It is **not** an additional charge to subtract from actual contract-level futures P&L.

For the matched-sleeve 60/40 portfolio:

$$
R_{60/40}=0.60R_E+0.40R_B
$$

so, before costs:

$$
R_{90/60}^{\mathrm{gross}}-R_f
=1.5(R_{60/40}-R_f)
$$

This identity holds for periodic arithmetic returns under its stated assumptions. It does not imply equal CAGR, equal drawdown, unchanged volatility under dynamic weights, or identical outcomes when futures, collateral currencies and costs differ.

### 8.4 Financing, basis and costs

The practical proxy adds explicit, non-overlapping frictions:

$$
R_{90/60,t}^{\mathrm{net}}
=0.90R_{E,t}+0.60R_{B,t}-0.50R_{f,t}
-C_{\mathrm{basis},t}
-C_{\mathrm{roll},t}
-C_{\mathrm{trade},t}
-C_{\mathrm{fund},t}
$$

For an annualized financing/basis spread $s_{\mathrm{basis},t}$ and period fraction $\Delta t$:

$$
C_{\mathrm{basis},t}=0.60\,s_{\mathrm{basis},t}\Delta t
$$

This is a scenario input, not a claim that all futures pay a constant spread. It captures a wedge between the assumed cash-index excess return and implementable futures economics.

For rebalance or roll transaction costs, define traded notional $T_t$ and a one-way proportional cost $k_t$ containing half-spread, commission, fees and estimated market impact:

$$
C_{\mathrm{trade},t}=k_t\frac{T_t}{V_{t-1}}
$$

A scheduled futures roll replaces an expiring contract with the next eligible contract. The price difference between contracts is not automatically a loss: it reflects carry, curve shape and time to maturity. The portfolio's economic roll return belongs in a verified futures-return series. In the proxy, use a separately disclosed roll-execution cost only:

$$
C_{\mathrm{roll},t}=k_{\mathrm{roll},t}\frac{T_{\mathrm{roll},t}}{V_{t-1}}
$$

Never subtract both a roll return embedded in $R_B$ and another assumed roll loss for the same effect. The NTSG-inspired proxy has no historical contract series, so $C_{\mathrm{roll}}$ is an execution scenario, not measured NTSG roll performance.

Fund cost is applied once. If $f$ is an annual fund fee over a period fraction $\Delta t$:

$$
1-C_{\mathrm{fund},t}=(1-f)^{\Delta t}
$$

Use this only for an index-based proxy whose inputs are not already net of the same fund fee. ETF adjusted-close returns already include fund expenses and distributions.

### 8.5 Margin and deleveraging stress test

Let $h_{j,t}$ be an initial-margin fraction. Required initial margin is:

$$
\mathrm{IM}_t=\sum_j h_{j,t}\left|q_{j,t}M_jF_{j,t}X_{j,t}\right|
$$

Let $A_t$ be liquid collateral after variation margin and costs. A transparent stress policy requires all of the following before it is run:

1. a stated initial-margin schedule $h_{j,t}$;
2. a variation-margin convention and settlement frequency;
3. a collateral floor $A_{\min}$; and
4. a forced-deleveraging rule if $A_t<A_{\min}$.

For example, a proportional notional reduction factor can be defined as:

$$
\lambda_t=\min\left(1,\frac{A_t-A_{\min}}{\mathrm{IM}_t-A_{\min}}\right)
$$

with a valid denominator and $0\leq\lambda_t\leq1$. The next-period futures target becomes $0.60\lambda_t$ rather than $0.60$. This is a model rule, not NTSG's rule.

Monthly data permits a **month-end collateral stress test** only. It cannot identify intra-month variation-margin calls, liquidity gaps or the exact timing of a forced sale. Daily settlement prices, historical margin schedules and cash-rate data are required before estimating historical liquidation frequency.

### 8.6 Required reporting

Every result must state whether it uses:

- the gross cash-index identity;
- a net proxy with the full cost grid; or
- a margin-stress scenario.

Report each cost component, gross risky exposure, cash allocation, financing reference, rebalance rule, bond currency policy, date range and currency. Compare the NTSG-inspired proxy with identical-sleeve 60/40 and 100% equity benchmarks; reserve claims about NTSG itself for the post-November-2024 live validation period.
